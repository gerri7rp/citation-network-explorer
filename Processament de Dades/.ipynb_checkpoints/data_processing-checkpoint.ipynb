```python
def none_if_empty(value):
    return None if value == "" else value
```

```python
def guardar_batch(data, num):
    df = pd.DataFrame(data)
    df.to_csv(f"datos_parciales_{num}.csv", index=False)

batch_size = 100_000
batch_num = 0
data = []
```

```python
#5,259,858 papers and 36,630,661 citation relationships
data = []

with open("../Data/dblp_v14.json", "r", encoding="utf-8") as file:
    objetos = ijson.items(file, "item")

    for i, obj in enumerate(objetos):   
        if i % 200_000 == 0:
            print(i)
        
        
        if i % 200_000 == 0 and i > 0 :
            guardar_batch(data, batch_num)
            batch_num += 1
            data = []



            # Obtenemos los datos
            id_                  = obj["id"]
            title                = obj["title"]
            authorsName          = [none_if_empty(author["name"]) for author in obj["authors"]] if "authors" in obj else []
            authorsOrganization  = [none_if_empty(author["org"])  for author in obj["authors"]] if "authors" in obj else []
            authorsId            = [none_if_empty(author["id"])   for author in obj["authors"]] if "authors" in obj else []
            venue                = obj["venue"]["raw"] if "venue" in obj and "raw" in obj["venue"] else ""
            year                 = obj["year"]
            keywords             = obj["keywords"]
            fieldOfStudy         = [fos["name"] for fos in obj["fos"]] if "fos" in obj else []
            weightOfFieldOfStudy = [fos["w"]    for fos in obj["fos"]] if "fos" in obj else []
            references           = obj.get("references", []) 
            n_citation           = obj["n_citation"]
            page_start           = obj["page_start"]
            page_end             = obj["page_end"]
            doc_type             = obj["doc_type"]
            lang                 = obj["lang"]
            volume               = obj["volume"]
            issue                = obj["issue"]
            issn                 = obj["issn"]
            isbn                 = obj["isbn"]
            doi                  = obj["doi"]
            url                  = ", ".join(obj["url"]) if "url" in obj else ""
            abstract             = obj["abstract"]
            #indexed abstract

            # Guardamos los datos en la lista
            data.append({
                "id": id_,
                "title": title,
                "authorsName": authorsName,
                "authorsOrganization" : authorsOrganization,
                "authorsId" : authorsId,
                "venue": venue,
                "year": year,
                "keywords" : keywords,
                "fieldOfStudy" : fieldOfStudy,
                "weightOfFieldOfStudy" : weightOfFieldOfStudy,
                "references" : references,
                "n_citation": n_citation,
                "page_start": page_start,
                "page_end": page_end,
                "doc_type": doc_type,
                "lang": lang,
                "volume": volume,
                "issue": issue,
                "issn" : issn,
                "isbn": isbn,
                "doi": doi,
                "url": url,
                "abstract": abstract,
            })
            
            
            print(title)
            

```

```python
import pandas as pd
import ast
from decimal import Decimal

# Columnas
columnas_lista = ["references", "authorsName", "authorsOrganization", "authorsId", "fieldOfStudy"]
col_decimal = "weightOfFieldOfStudy"

# Funciones optimizadas
def parse_lista(celda):
    if pd.isna(celda) or celda.strip() == "":
        return []
    try:
        return ast.literal_eval(celda)
    except:
        return []

def parse_decimales(celda):
    if pd.isna(celda) or celda.strip() == "":
        return []
    try:
        return [float(x) for x in eval(celda, {"Decimal": Decimal})]
    except:
        return []

# Lectura por chunks
ruta_csv = "completo.csv"
chunksize = 25000
df_final = pd.DataFrame()

count = 0
for chunk in pd.read_csv(ruta_csv, chunksize=chunksize):
    # Solo procesar columnas necesarias
    for col in columnas_lista:
        if col in chunk.columns:
            chunk[col] = chunk[col].map(parse_lista)
    
    if col_decimal in chunk.columns:
        chunk[col_decimal] = chunk[col_decimal].map(parse_decimales)
    
    df_final = pd.concat([df_final, chunk], ignore_index=True)
    
    count += chunksize
    print(count)


print("Finalizado:", len(df_final), "filas")
```

```python
import ast
import pandas as pd
from decimal import Decimal

# Columnas que deben ser listas
columnas_como_lista = ["references", "authorsName", "authorsOrganization", "authorsId", "fieldOfStudy"]
columna_decimal = "weightOfFieldOfStudy"

# Ruta de archivo
ruta_csv = "completo.csv"
chunksize = 25000#0
df_final = pd.DataFrame()

def limpiar_lista_str(lista):
    """Convierte strings como "['a', 'b']" → ['a', 'b'], y limpia comillas internas."""
    try:
        l = ast.literal_eval(lista)
        if isinstance(l, list):
            return [s.strip("'\"") if isinstance(s, str) else s for s in l]
        return []
    except:
        return []

def parsear_decimales(cadena):
    """Convierte string con Decimals en lista de floats."""
    try:
        lista = eval(cadena, {"Decimal": Decimal})
        if isinstance(lista, list):
            return [float(x) for x in lista]
        return []
    except:
        return []

# Leer en chunks y procesar
count = 0
for chunk in pd.read_csv(ruta_csv, chunksize=chunksize):
    # Parsear las columnas normales
    chunk.loc[:, columnas_como_lista] = chunk[columnas_como_lista].applymap(limpiar_lista_str)
    
    # Parsear la columna con Decimals
    chunk[columna_decimal] = chunk[columna_decimal].apply(parsear_decimales)
    
    df_final = pd.concat([df_final, chunk], ignore_index=True)
    count += chunksize
    print(count)
    break
```

Se ordenan los articulos por numero de vecinos

```python
df_cc['neighbors_count'] = df_cc['neighbors'].str.len()
df_cc["Component"] = None
df_cc = df_cc.set_index("id")
```

```python
df_cc = df_cc.sort_values(by='neighbors_count', ascending=False)
```

```python
df_cc.head(50)
```

```python
components_count = 0

all_not_visited_nodes = df_cc.loc[df_cc["Component"].isna()]

while all_not_visited_nodes.shape[0] > 0:
    node_id = all_not_visited_nodes.index[0]
    component_id = df_cc.loc[node_id, "Component"]
    print("\nNew Component:", components_count)
    print(df_cc.loc[node_id, "neighbors_count"])

    # Si no esta asignado a un component id
    if component_id == None: 
        df_cc.loc[node_id, "Component"] = components_count    
        neighbors = [node_id] 

        while neighbors:
            # Lista de todos los vecinos de los vecinos
            neighbors = df_cc.loc[neighbors, "neighbors"]
            neighbors = neighbors.explode().dropna().to_list()

            # Quitamos repetidos y ids no existentes
            neighbors = list(set(neighbors) & set(df_cc.index))

            # Obtenemos que vecinos no tienen asignado un componente id
            mask = df_cc.loc[neighbors, "Component"].isna()
            mask_neighs_none = df_cc.loc[neighbors].index[mask]
            
            # Solo cojemos una muestra de x
            mask_neighs_none = mask_neighs_none[:100]
            print(mask_neighs_none.shape)

            # Asignamos el componente id a los nodos que no tenian uno previamente
            df_cc.loc[mask_neighs_none, "Component"] = components_count

            # Guardamos solo nos vecinos no visitados
            neighbors = mask_neighs_none.tolist()

            print("Total nodos recorridos:", df_cc["Component"].notna().sum())

            if df_cc["Component"].notna().sum() >= 1_000:
                break
            
    components_count += 1
    all_not_visited_nodes = df_cc.loc[df_cc["Component"].isna()]
    
    # Solo queremos el primer componente
    break
```
